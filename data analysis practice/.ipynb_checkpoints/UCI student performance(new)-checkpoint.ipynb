```python
import numpy as np
import pandas as pd
from ucimlrepo import fetch_ucirepo
```

```python
student = fetch_ucirepo(id=320)
```

```python
df = student.data.original.copy()
df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df_numeric = df.select_dtypes(include='number').columns.tolist()
```

```python
df_numeric
```

```python
import matplotlib.pyplot as plt
```

```python
for feature in df_numeric:
    plt.figure(figsize=(5,5))
    plt.title(f'distribution of {feature}')
    plt.xlabel(feature)
    plt.ylabel('frequency')
    df[feature].hist(bins=20,color='skyblue',edgecolor='black')
    plt.show()
```

```python
df_objects = df.select_dtypes(include='object').columns.tolist()
df_objects
```

```python
for feature in df_objects:
    print(f'{feature}: {df[feature].unique()}')
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
del df_objects[5:9]
for feature in df_objects:
    df[feature] = le.fit_transform(df[feature])
```

```python
ohe_objects = ['Mjob','Fjob','reason','guardian']
for feature in ohe_objects:
    df = pd.get_dummies(df,columns=[feature])
```

```python
df.info()
```

```python
import seaborn as sns
plt.figure(figsize=(15,8))
sns.heatmap(df.corr(), annot=True, fmt='.2f', cmap='coolwarm',annot_kws={'size':5})
plt.show()
```

```python
df.describe()
```

```python
X = df.drop('G3', axis=1)
y = df['G3']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=.2, random_state = 17)
```
